```python
X = df.drop(columns=["target", "ID"])
y = df["target"]
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.3, random_state=42, stratify=y)
```

```python
y_train.value_counts(normalize=True)*100
```

```python
y_test.value_counts(normalize=True)*100
```

### Entrenamiento y evaluacion de modelos

#### Arbol de Decision

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
decision_tree = DecisionTreeClassifier(max_depth=4, random_state=42)
```

```python
%%time
decision_tree.fit(X_train, y_train)
```

```python
y_pred_train_dt = decision_tree.predict(X_train) # predict -> devuelve las clases: 0 o 1
y_pred_test_dt = decision_tree.predict(X_test)
```

```python
y_pred_train_dt
```

```python
y_train
```

```python
from sklearn.metrics import classification_report
```

```python
# No es trivial la eleccion del mejor modelo en base a su performance, depende siempre del contexto de negocio y nuestro objetivo
```

```python
# IMPORTANTE: SIEMPRE EVALUAR LA PERFORMANCE DE UN MODELO ENTRENADO POR UNO MISMO EN TODOS LOS CONJUNTOS
print("TRAIN")
print(classification_report(y_train, y_pred_train_dt))
print("--------------------------------------------------------")
print("TEST")
print(classification_report(y_test, y_pred_test_dt))
```

```python
# IMPORTANTE: SIEMPRE EVALUAR LA PERFORMANCE DE UN MODELO ENTRENADO POR UNO MISMO EN TODOS LOS CONJUNTOS
# performance con max_depth = 1 -> seguramente uso "gender" para predecir
print("TRAIN")
print(classification_report(y_train, y_pred_train_dt))
print("--------------------------------------------------------")
print("TEST")
print(classification_report(y_test, y_pred_test_dt))
```
